```python
import pandas as pd
```

```python
# загружаем датасет
data = pd.read_csv("./data/adult.data.csv")
data.head()
```

**1. Посчитайте, сколько мужчин и женщин (признак *sex*) представлено в этом датасете**

**2. Каков средний возраст мужчин (признак *age*) по всему датасету?**

```python
mean_age = round(data['age'].mean(), 2)
print(mean_age)
```

**3. Какова доля граждан Соединенных Штатов (признак *native-country*)?**

```python
country_column = 'native-country'
target_country = 'United-States'

country_proportions = data[country_column].value_counts(normalize=True)
us_proportion = country_proportions.get(target_country, 0) * 100

print(f"Доля граждан Соединённых Штатов: {round(us_proportion, 2)}")
```

**4-5. Рассчитайте среднее значение и среднеквадратичное отклонение возраста тех, кто получает более 50K в год (признак *salary*) и тех, кто получает менее 50K в год**

```python
more_than_fifty_k = data.loc[data['salary'] == '>50K', 'age']
less_than_fifty_k = data.loc[data['salary'] == '<=50K', 'age']

print(f"Среднее значение возраста тех, кто получает больше 50K: {round(more_than_fifty_k.mean(), 2)}")
print(f"Среднее значение возраста тех, кто получает меньше 50K: {round(less_than_fifty_k.mean(), 2)}")

print(f"Среднеквадратичное отклонение возраста тех, кто получает больше 50K: {round(more_than_fifty_k.std(), 2)}")
print(f"Среднеквадратичное отклонение возраста тех, кто получает меньше 50K: {round(less_than_fifty_k.std(), 2)}")
```

**6. Правда ли, что люди, которые получают больше 50k, имеют минимум высшее образование? (признак *education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* или *Doctorate*)**

**7. Выведите статистику возраста для каждой расы (признак *race*) и каждого пола. Используйте *groupby* и *describe*. Найдите таким образом максимальный возраст мужчин расы *Asian-Pac-Islander*.**

**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак *marital-status*)? Женатыми считаем тех, у кого *marital-status* начинается с *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
is_married = data['marital-status'].str.startswith('Married')
is_not_married = ~is_married
is_male = (data['sex'] == 'Male')
is_high_salary = (data['salary'] == '>50K')

mask_married_males_high_salary = is_married & is_male & is_high_salary
count_married_males_high_salary = data.loc[mask_married_males_high_salary, 'sex'].count()

mask_not_married_males_high_salary = is_not_married & is_male & is_high_salary
count_unmarried_males_high_salary = data.loc[mask_not_married_males_high_salary, 'sex'].count()

if count_married_males_high_salary > count_unmarried_males_high_salary:
    print("Количество женатых мужчин, зарабатывающих более 50К, больше, чем количество холостых мужчин")
    print(f"Количество женатых мужчин: {count_married_males_high_salary}")
    print(f"Количество неженатых мужчин: {count_unmarried_males_high_salary}")
elif count_married_males_high_salary < count_unmarried_males_high_salary:
    print("Количество женатых мужчин, зарабатывающих более 50К, меньше, чем количество холостых мужчин")
    print(f"Количество женатых мужчин: {count_married_males_high_salary}")
    print(f"Количество неженатых мужчин: {count_unmarried_males_high_salary}")
else:
    print("Количество женатых мужчин, зарабатывающих более 50К, такое же, как и количество холостых мужчин")
    print(f"Количество женатых мужчин: {count_married_males_high_salary}")
    print(f"Количество неженатых мужчин: {count_unmarried_males_high_salary}")
```

**11.Сгруппируйте людей по возрастным группам *young*, *adult*, *retiree*, где:**
* *young* соответствует 16-35 лет
* *adult* - 35-70 лет
* *retiree* - 70-100 лет

**Проставьте название соответсвтуещей группы для каждого человека в новой колонке AgeGroup**

```python
bins = [16, 35, 70, 100]
labels = ('young', 'adult', 'retiree')

data['age-group'] = pd.cut(
    x=data['age'],
    bins=bins,
    labels=labels,
    right=False
)

print(data)
```

**12-13. Определите количество зарабатывающих >50K в каждой из возрастных групп (колонка AgeGroup), а также выведите название возрастной группы, в которой чаще зарабатывают больше 50К (>50K)**

```python
is_high_salary = data['salary'] == '>50K'
age_groups = data.loc[is_high_salary].groupby('age-group')['salary'].count()

count_high_salary_per_group = (data['salary'] == '>50K').groupby(data['age-group']).sum()

top_group = count_high_salary_per_group.idxmax()
max_count = count_high_salary_per_group.max()

print(age_groups)
print(f"Чаще всего больше 50К зарабатывает группа {top_group}")
print(f"Количество человек в этой группе: {max_count}")
```

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
def filter_func(x):
    mean_age = x['age'].mean() <= 40
    all_hours_condition = (x['hours-per-week'] > 5).all()
    return mean_age and all_hours_condition

occupation_size = data.groupby('occupation').size()
print("Группировка людей по занятости и иформация о количестве в каждой группе:")
print(occupation_size)

filtered_df = data.groupby('occupation').filter(filter_func)
print("\nОтфильтрованный DataFrame:\n")
print(filtered_df)

print("\nОставшиеся группы:\n")
print(filtered_df['occupation'].unique())
```

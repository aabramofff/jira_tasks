* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
# without numpy
x = [[1, 0, 1], [2, 4, 2], [3, 0, 3], [4, 4, 4]]
res1 = 1

for i in range(len(x[0])):
    res1 *= x[i][i] if x[i][i] != 0 else 1

print(f'Result using only Python: {res1}')

# using NumPy
x = np.array([[1, 0, 1], [2, 4, 2], [3, 0, 3], [4, 4, 4]])
n = 1

diagonal = np.diagonal(x)
filtered_diagonal = diagonal[np.where(diagonal != 0)]
res2 = np.prod(filtered_diagonal)

print(f'Result using only NumPy: {res2}')
```

* __Задача 2__: Даны два вектора x и y. Проверить, задают ли они одно и то же мультимножество.  
  Например, для x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) ответ True.

```python
# using Python
x = [1, 2, 2, 4]
y = [4, 2, 1, 2]
n_x = len(x)
n_y = len(y)
res1 = n_x == n_y

if res1:
    x_unique, y_unique = set(x), set(y)
    x_count, y_count = [x.count(el) for el in x_unique], [y.count(el) for el in y_unique]

    for i in range(len(x_unique)):
        res = x_unique == y_unique and x_count == y_count
        if res:
            continue
        else:
            break

print(f'Result using Python: {res1}')

# using NumPy
x = np.array([1, 2, 2, 4])
y = np.array([4, 2, 1, 2])

x_unique, x_count = np.unique(x, return_counts=True)
y_unique, y_count = np.unique(y, return_counts=True)

res2 = np.array_equal(x_unique, y_unique) and np.array_equal(x_count, y_count)

print(f'Result using NumPy: {res2}')
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль.
 Например, для x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) ответ 5.

```python
# using Python
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]
max = x[1]
res1 = None

for i in range(1, len(x)):
    if x[i-1] == 0 and x[i] > max:
        res1 = x[i]
        max = x[i]

print(f'Result using Python: {res1}')

# using NumPy
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])

mask_zeros = x[:-1] == 0

elements_after_zeros = x[1:][mask_zeros]

if elements_after_zeros.size == 0:
    res2 = None
else:
    res2 = np.max(elements_after_zeros)

print(f'Result using NumPy: {res2}')
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import numpy as np
import timeit
from scipy.spatial.distance import cdist

# --- Генерация массивов ---
N_X = 5
N_Y = 4
D = 3
np.random.seed(42)
X_np = np.random.uniform(0, 10, size=(N_X, D)).round(2)
Y_np = np.random.uniform(0, 10, size=(N_Y, D)).round(2)
X_list = X_np.tolist()
Y_list = Y_np.tolist()


# --- Метод 1: Чистый Python (Вложенные циклы) ---
def euclidean_distance_python(X, Y):
    N_X = len(X)
    N_Y = len(Y)
    D_matrix = [[0.0] * N_Y for _ in range(N_X)]
    
    for i in range(N_X):
        for j in range(N_Y):
            sum_sq = 0.0
            for k in range(len(X[i])):
                diff = X[i][k] - Y[j][k]
                sum_sq += diff * diff
            D_matrix[i][j] = sum_sq ** 0.5
            
    return np.array(D_matrix)


# --- Метод 2: Только NumPy (Векторизация) ---
def euclidean_distance_numpy(X_np, Y_np):
    X_sq_norms = np.sum(X_np**2, axis=1)[:, np.newaxis]
    Y_sq_norms = np.sum(Y_np**2, axis=1)[np.newaxis, :]
    XY_dot = -2 * np.dot(X_np, Y_np.T)
    D_sq = X_sq_norms + XY_dot + Y_sq_norms
    D_sq[D_sq < 0] = 0
    
    return np.sqrt(D_sq)


# --- Проверка и сравнение ---
print("--- Исходные массивы ---")
print(f"X:\n{X_np}")
print(f"\nY:\n{Y_np}")

# Проверка корректности (должны быть идентичны)
D_py = euclidean_distance_python(X_list, Y_list)
D_np = euclidean_distance_numpy(X_np, Y_np)
D_scipy = cdist(X_np, Y_np, metric='euclidean')

print("\n--- Проверка корректности (Max разница) ---")
print(f"Python vs NumPy: {np.max(np.abs(D_py - D_np)):.2e}")
print(f"NumPy vs SciPy:  {np.max(np.abs(D_np - D_scipy)):.2e}")

# Измерение времени (используем более крупный массив для значимого сравнения)
N_BIG = 1000
D_BIG = 10
X_BIG = np.random.rand(N_BIG, D_BIG)
Y_BIG = np.random.rand(N_BIG, D_BIG)
X_BIG_list = X_BIG.tolist()
Y_BIG_list = Y_BIG.tolist()

num_runs = 5

time_py = timeit.timeit(lambda: euclidean_distance_python(X_BIG_list, Y_BIG_list), number=1) * 1e3
time_np = timeit.timeit(lambda: euclidean_distance_numpy(X_BIG, Y_BIG), number=num_runs) / num_runs * 1e3
time_scipy = timeit.timeit(lambda: cdist(X_BIG, Y_BIG, metric='euclidean'), number=num_runs) / num_runs * 1e3

print("\n--- Сравнение производительности (1000x1000x10) ---")
print(f"1. Чистый Python (1 запуск):    {time_py:.2f} мс")
print(f"2. NumPy (векторизация):        {time_np:.2f} мс")
print(f"3. SciPy cdist (оптимизация C): {time_scipy:.2f} мс")
print("-" * 40)
print(f"NumPy быстрее Python примерно в {time_py / time_np:.0f} раз")
```
